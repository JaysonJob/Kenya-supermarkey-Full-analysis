```python
import pandas as pd
from scipy import stats
```

Runing the Statistical Test

```python
# Testing if M Pesa users spend more than Credit Card users
mpesa_spend = df[df['payment_method'] == 'M-Pesa']['total_amount']
card_spend = df[df['payment_method'] == 'Credit Card']['total_amount']
```

```python
# Run the T-test
t_stat,p_value = stats.ttest_ind(mpesa_spend,card_spend,equal_var=False)

print(f"P-value: {p_value}")
```

Because P-value is 0.88 (which is much higher than 0.05), it means there is no significant difference in spending between the two groups.
